```python
import numpy as np
import matplotlib.pyplot as plt

#Data structure:[restaurant_id ,2021,2022,2023,2024]
sales_data = np.array([
    [1,120, 150, 130, 170, 160],  #paradise birayani
    [2,200, 210, 190, 220, 205],  # Beijing Bites
    [3,90,  110, 105, 115, 100],  #Pizza Hub
    [4,300, 320, 310, 330, 340],  #Burger Bos
    [5,180, 175, 185, 190, 200]   #Chai code
])

print(sales_data.shape)

print(sales_data)
```

```python
#data of first three restorant
print(" Sample data for 1st 3 restau",sales_data[:3,1:])
#total sale per year
print(np.sum(sales_data ,axis=0)) #if axis 0 sum column wise
#minimum sales per restaurant 
print("min salses per restaurant:", np.min(sales_data[:,3:],axis=1))

#maximum sale per year
print("maximum sale per year",np.max(sales_data[:,1:] ,axis=0))

#average sale per restautant
avg_sale=np.mean(sales_data[:,1:],axis=1)
print("Avg_sale",avg_sale)

#cumulative sales
cumsum=np.cumsum(sales_data[:,1:],axis=1)
print("cumsum",cumsum)
print("cumsum mean",np.mean(cumsum[:,:],axis=0))

   
```

1. Vectorization = How the operation is executed

Vectorization means NumPy performs operations on the entire array at once using optimized C code instead of a Python loop.

2. Broadcasting = How shapes are made compatible

Broadcasting means NumPy automatically expands a smaller array (or scalar) to match a larger array's shape.
