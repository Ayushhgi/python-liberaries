```python
import numpy as np

sales_data = np.array([
    [120, 150, 130, 170, 160],  
    [200, 210, 190, 220, 205], 
    [90,  110, 105, 115, 100],  
    [300, 320, 310, 330, 340],
    [180, 175, 185, 190, 200]   
])

print(sales_data.shape)

print(sales_data)
```

```python
#total sale per year
print("Total sales:" ,np.sum(sales_data[:,1:],axis=0)) # if axis=0 sum will be column wise
#minimum sales per restaurant 
print("min salses per restaurant:", np.min(sales_data[:,3:],axis=1))
```
